```python
import torch
from torch.nn.functional import log_softmax, softmax
```

```python
a = torch.tensor([1, 2])
e = torch.tensor([
    [
        2,5
    ],
    [
        6,7
    ]
])
b = torch.tensor([
    [
        4
    ],
    [5]
])
```

```python
a @ e
```

```python
b.T * (a @ e)
```

```python
eLog = torch.log(e)
aLog = torch.log(a)
```

```python
summed = aLog + eLog.T
```

```python
summed
```

```python
torch.exp(summed)
```

```python
torch.exp(torch.logsumexp(summed, dim=-1))
```

```python
logAE = torch.logsumexp(summed, dim=-1)
logB = torch.log(b)
```

```python
torch.exp(logB.T + logAE)
```
