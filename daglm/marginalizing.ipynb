```python
import torch
```

```python
targetLen = 4
lambdaMult = 2
graphLen = targetLen * lambdaMult
vocabSize = 5
```

```python
tokenLogits = torch.randn(graphLen, vocabSize)
transitionLogits = torch.randn(graphLen, graphLen)
```

```python
from torch.nn.functional import log_softmax
```

```python
tokenLogProbs = log_softmax(tokenLogits, dim=-1)
```

```python
# create lower triangular mask for transition logits
mask = torch.tril(torch.ones(graphLen, graphLen))
```

```python
mask
```

```python
transitionLogits = transitionLogits.masked_fill(mask == 1, float('-inf'))
```

```python
transitionLogProbs = log_softmax(transitionLogits, dim=-1)
```

```python
transitionLogProbs
```

```python
dP = torch.zeros(targetLen, graphLen)
```

```python
dP[0,0] = 1
```

```python
i = 1
```

```python
dP[i:] = tokenLogProbs
```
