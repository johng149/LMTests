```python
import torch
```

```python
seq_len = 5
batch_size = 3
vocab_size = 6 # 0-5, actual vocab size is 7 when including padding
pad_idx = 6
seq_len_padded = seq_len + 2
```

```python
tokens = []
for i in range(batch_size):
    sampled_len = torch.randint(1, seq_len, (1,)).item()
    tokens.append(torch.randint(1, vocab_size, (sampled_len,)))
```

```python
tokens
```

```python
tokens = torch.nested.nested_tensor(tokens)
```

```python
tokens = torch.nested.to_padded_tensor(tokens, pad_idx, (batch_size, seq_len_padded))
```

```python
tokens
```

```python
not_pad = tokens != pad_idx
```

```python
not_pad
```

```python
# find all columns in `not_pad` that sum to 0 (i.e. all padding columns)
pad_cols = torch.where(not_pad.sum(dim=0) == 0)[0]
```

```python
pad_cols
```

```python
not_pad.any(dim=0)
```

```python
# remove all padding columns
tokens = tokens[:, not_pad.any(dim=0)]
```
