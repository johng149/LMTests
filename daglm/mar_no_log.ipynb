```python
import torch
```

```python
targetLen = 4
lambdaMult = 2
graphLen = targetLen * lambdaMult
vocabSize = 5
```

```python
tokenLogits = torch.randn(graphLen, vocabSize)
transitionLogits = torch.randn(graphLen, graphLen)
```

```python
mask = torch.tril(torch.ones(graphLen, graphLen))
```

```python
transitionLogits = transitionLogits.masked_fill(mask == 1, float('-inf'))
```

```python
from torch.nn.functional import softmax
```

```python
tokenProbs = softmax(tokenLogits, dim=-1)
transitionProbs = softmax(transitionLogits, dim=-1)
```

```python
dP = torch.zeros(targetLen, graphLen)
```

```python
dP[0, 0] = 1 # guaranteed to start at BOS
```

```python
i = 1
```

```python
dP[i-1, :] @ transitionProbs
```
