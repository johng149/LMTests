```python
import torch
```

```python
batch_size, seq_len, emb_dim = 2, 3, 4
vocab_size = 4
targets = torch.LongTensor([[0, 1, 3], [3, 1, 0]])
```

```python
vocab_classifier_head = torch.nn.Linear(emb_dim, vocab_size)
```

```python
test_hidden = torch.randn(batch_size, seq_len, emb_dim)
```

```python
out = vocab_classifier_head(test_hidden)
```

```python
out = torch.log_softmax(out, dim=-1)
```
