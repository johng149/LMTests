```python
import torch
```

```python
batch_size, seq_len, emb = 2, 3, 4
```

```python
x = torch.randn(batch_size, seq_len, emb)
```

```python
attn_proj = torch.nn.Linear(emb, emb*3)
```

```python
num_heads = 2
```

```python
q, k, v = attn_proj(x).split(emb, dim=-1)
```

```python
q = q.view(batch_size, -1, num_heads, emb//num_heads).transpose(1, 2)
k = k.view(batch_size, -1, num_heads, emb//num_heads).transpose(1, 2)
v = v.view(batch_size, -1, num_heads, emb//num_heads).transpose(1, 2)
```

```python
mask = torch.randn(batch_size, seq_len, seq_len) > 0
```

```python
mask.unsqueeze(1).shape
```

```python
from torch.nn.functional import scaled_dot_product_attention as sdpa
```

```python
sdpa(q, k, v, mask)
```
