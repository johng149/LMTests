```python
import torch
from torch import nn
```

```python
class SmallModel(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SmallModel, self).__init__()
        self.modules = nn.ModuleList(
            [nn.Linear(input_size, hidden_size), nn.ReLU(), nn.Linear(hidden_size, output_size)]
        )

    def forward(self, x):
        for module in self.modules:
            x = module(x)
        return x
```

```python
input_size = 10
hidden_size = 20
output_size = 5
```

```python
sm = SmallModel(input_size, hidden_size, output_size)
```

```python
print(f"Num parameters: {sum([p.numel() for p in sm.parameters()])}")
```

```python
class SmallModelNested(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SmallModelNested, self).__init__()
        self.layers = nn.ModuleList()
        for i in range(3):
            l1 = nn.Linear(input_size, hidden_size)
            l2 = nn.Linear(hidden_size, output_size)
            self.layers.append(nn.ModuleList([l1, l2]))
        
    def forward(self, x):
        for l1, l2 in self.layers:
            x = l2(torch.relu(l1(x)))
        return x
```

```python
smn = SmallModelNested(input_size, hidden_size, output_size)
```

```python
print(f"Num parameters: {sum([p.numel() for p in smn.parameters()])}")
```
