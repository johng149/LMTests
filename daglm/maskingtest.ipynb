```python
import torch
```

```python
batch, seq, emb = 2, 3, 4
```

```python
q = torch.rand(batch, seq, emb)
k = torch.rand(batch, seq, emb)
```

```python
qispad = torch.tensor(
    [
        [
            0, 1, 1
        ],
        [
            0, 0, 0
        ]
    ],
    dtype=torch.bool
)
assert qispad.shape == (batch, seq)
# qispad is a mask for q, where 1 means that the token is a padding token
kispad = torch.tensor(
    [
        [
            0, 0, 1
        ],
        [
            0, 0, 0
        ]
    ],
    dtype=torch.bool
)
assert kispad.shape == (batch, seq)
# kispad is a mask for k, where 1 means that the token is a padding token
```

```python
# we need a mask of shape (batch, seq, seq) not (batch, seq)
kispad = kispad.unsqueeze(1).expand(batch, seq, seq)
```

```python
kispad = ~kispad
```
