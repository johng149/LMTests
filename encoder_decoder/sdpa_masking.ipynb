```python
import torch
```

```python
pad_idx = -1
```

```python
test_input = torch.tensor([[1, 2, -1, -1], [5, 6, 7, -1], [9, 10, 11, 12]])
```

```python
batch_size, seq_len = test_input.size()
```

```python
attn_mask = (test_input != pad_idx).float()
```

```python
attn_mask
```

```python
attn_mask.masked_fill_(attn_mask.logical_not(), float('-inf'))
```

```python
attn_mask
```

```python
num_heads = 2
```

```python
attn_mask.masked_fill_(attn_mask == 1, 0)
```

```python
attn_mask
```

```python
attn_mask = attn_mask.unsqueeze(1).unsqueeze(2).repeat(1, num_heads, seq_len, 1)
```
