```python
import torch
```

```python
input_tensor = torch.tensor([[1, 2, 3, 3], [1, 2, 2, 3], [1, 1, 2, 2]])
pad_idx = 3
```

```python
mask = (input_tensor != pad_idx).float()
```

```python
mask
```

```python
mask.masked_fill_(mask.logical_not(), float('-inf'))
```

```python
mask.masked_fill_(mask == 1, 0)
```

```python
mask.unsqueeze(1).unsqueeze(2).repeat(1, 2, 2, 1)
```
